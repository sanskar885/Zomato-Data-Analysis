```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Zomato Project/Zomato data .csv')
df.head()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.head(2)
```

```python
df.info()
```

Conerting the data type of column rate

```python
def HandleRate(value):
    value = str(value).split('/')
    value = value[0]
    return float(value)

df['rate'] = df['rate'].apply(HandleRate)
df.head(2)
```

Restaurant with majority customers

```python
sns.countplot(x=df['listed_in(type)'])
plt.xlabel("Type of restaurant")
```

Votes recieved by each type of restaurant

```python
df.head(2)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

grouped_data = df.groupby('listed_in(type)')['votes'].sum().reset_index()
result = pd.DataFrame(grouped_data)

plt.figure(figsize=(10, 6))
plt.plot(result['listed_in(type)'], result['votes'], c='green', marker='o')
plt.xlabel("Type of Restaurant", color="red", fontsize=20)
plt.ylabel("Votes", color="red", fontsize=20)
plt.xticks(rotation=45)
plt.title("Votes per Restaurant Type", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.hist(df['rate'],bins=5)
plt.title("Rating Distribution")
plt.show()
```

Average spending of couples on each order

```python
couple_data=df['approx_cost(for two people)']
sns.countplot(x=couple_data)
```

Which mode has recieved maximum rating.(Online or Offline)

```python
plt.figure(figsize=(6,6))
sns.boxplot(x='online_order',y='rate',data=df)
```

Which restaurant have more offline order

```python
pivot_table = df.pivot_table(index='listed_in(type)', columns='online_order', values='votes', aggfunc='sum', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap="YlGnBu", fmt='d')
plt.title('Heat map')
plt.xlabel("Online Order")
plt.ylabel("listed_in(type)")
plt.show()
```
